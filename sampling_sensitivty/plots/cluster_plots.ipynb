How the clustering is done:

For each configuration (one feature type, one sampling strategy, one sample size) and each of the 30 runs: take that run's feature values for all 2400 instances to form a 2400 × n_features matrix. Standardize each feature column to zero mean and unit variance. Apply PCA retaining 99% of explained variance to reduce dimensionality. Run k-means with k=24 (matching the 24 BBOB function classes) and 50 random initializations, keeping the best result. This produces a predicted cluster label for each of the 2400 instances.

ARI — Adjusted Rand Index:
The Rand Index considers every pair of instances (about 2.9 million pairs for 2400 instances) and counts how often the clustering agrees with the true labels — either both say "same group" or both say "different group." The adjustment corrects for agreement that would occur by random chance.
ARI = 1 means the clustering perfectly recovers the true class structure. ARI = 0 means the clustering is no better than random assignment. ARI can go slightly negative, meaning worse than random. In your context, a high ARI means the features place instances of the same function class together and instances of different classes apart.

NMI — Normalized Mutual Information:
Mutual information quantifies how much information the cluster assignments share with the true labels — how much your uncertainty about which class an instance belongs to is reduced by knowing which cluster it was assigned to. The normalization scales it between 0 and 1 so it's comparable across different numbers of clusters.
NMI = 1 means knowing the cluster tells you the class with certainty, and vice versa. NMI = 0 means the clusters and classes are statistically independent — knowing one tells you nothing about the other.
