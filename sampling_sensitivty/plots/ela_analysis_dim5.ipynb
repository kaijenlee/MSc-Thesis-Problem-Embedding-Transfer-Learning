## ELA Features NANs

```python
dim = 5
current_case = "cma_random_25"
current_run = (1,1,dim)
current_data_dir = ela_files_dict[current_case]

with open(current_data_dir, "rb") as f:
    current_dataset = pickle.load(f)

nan_dict_base = {}
for current_feat in ela_feature_keys:
    print(f"# {current_feat}")
    for key in current_dataset[current_run][current_feat][0].keys():
        print(f"{key}: {current_dataset[current_run][current_feat][0][key]}")
        nan_dict_base[key] = {"nan": [], "inf":[]}
    print()
```

```python
nan_dict_base
```

```python
def get_ela_invalid(current_dataset, dim=2):
    current_nan = copy.deepcopy(nan_dict_base)
    levelset_failed = []
    for function in range(1,25):
        for instance in range(1,101):
            current_run = (function,instance,dim)
            for current_feat in ela_feature_keys:
                for run_num in range(30):
                    if current_feat == "levelset" and len( current_dataset[current_run][current_feat][run_num]) == 0:
                        levelset_failed.append(current_run + (run_num,))
                        continue
                    for key in current_dataset[current_run][current_feat][run_num].keys():
                        val = current_dataset[current_run][current_feat][run_num][key]
                        if np.isnan(val):
                            current_nan[key]["nan"].append(current_run + (run_num,))
                        elif np.isinf(val):
                            current_nan[key]["inf"].append(current_run + (run_num,))
    affected_keys = {k for k,v in current_nan.items() if k != "levelset_failed" and v['nan'] or v['inf']}
    filtered = {k: v for k, v in current_nan.items() if  k != "levelset_failed" and v['nan'] or v['inf']}
    return affected_keys, filtered, levelset_failed
```

```python
dim = 5
current_case = "cma_random_25"
current_run = (1,1,dim)
current_data_dir = ela_files_dict[current_case]

with open(current_data_dir, "rb") as f:
    current_dataset = pickle.load(f)

affected_keys, filtered, _ = get_ela_invalid(current_dataset, dim)
affected_keys
```

```python
nan_summary = {}
set_overall = set()
for settings in ela_files_dict.keys():
    print(f"Case: {settings}")
    with open(ela_files_dict[settings], "rb") as f:
        current_dataset = pickle.load(f)
    affected_keys, filtered, failed_levelset = get_ela_invalid(current_dataset, dim)
    set_overall.update(affected_keys)
    print(f"Affected Keys: {affected_keys}, Failed Levelset: {len(failed_levelset)}")
    nan_summary[settings] = {"affected_keys": affected_keys, "filtered": filtered, "failed_levelset": failed_levelset}
    print()

print(f"Overall Affected Keys: {set_overall}")
```

```python
with open(data_dir / "ela_nan_inf_summary.pkl", "wb") as f:
    pickle.dump(nan_summary, f)
```
